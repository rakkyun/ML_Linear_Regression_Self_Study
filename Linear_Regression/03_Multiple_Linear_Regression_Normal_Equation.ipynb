# 03 - Multiple Linear Regression using the Normal Equation

## Aim
Implement **Multiple Linear Regression** with three predictors using the normal equation. This demonstrates how the one-variable regression idea is extended to a matrix form.

### Model
For three predictors:

**ŷ = β₀ + β₁x₁ + β₂x₂ + β₃x₃**

In matrix notation the coefficients can be estimated by:

**β = (XᵀX)⁻¹Xᵀy**

Here a pseudo-inverse is used in code so the calculation is numerically safer.

## 1. Import libraries

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

## 2. Load the input and output data

```python
features = pd.read_csv("datasets/LR_X.csv")
target = pd.read_csv("datasets/LR_Y.csv")["Y"]

print("Feature matrix:", features.shape)
print("Target vector :", target.shape)

display(features.head())
display(target.head())
```

## 3. Add the intercept column
The first column contains ones so that β₀ can be learned together with the other coefficients.

```python
X = features.to_numpy(dtype=float)
y = target.to_numpy(dtype=float)

X_design = np.column_stack((np.ones(X.shape[0]), X))
print("Design matrix shape:", X_design.shape)
```

## 4. Calculate the coefficients

```python
beta = np.linalg.pinv(X_design.T @ X_design) @ X_design.T @ y

coef_names = ["β0", "β1", "β2", "β3"]
for name, value in zip(coef_names, beta):
    print(f"{name} = {value:.4f}")
```

## 5. Write the fitted equation

```python
b0, b1, b2, b3 = beta
print(f"ŷ = {b0:.2f} + ({b1:.2f})x1 + ({b2:.2f})x2 + ({b3:.2f})x3")
```

## 6. Predict the target values

```python
y_pred = X_design @ beta

preview = pd.DataFrame({
    "Actual": y,
    "Predicted": y_pred,
    "Residual": y - y_pred
})
preview.head(10)
```

## 7. Evaluate the model

```python
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"MAE : {mae:.8f}")
print(f"MSE : {mse:.8f}")
print(f"RMSE: {rmse:.8f}")
print(f"R²  : {r2:.4f}")
```

## 8. Predict a new observation

```python
new_observation = np.array([[1, 0.30, 0.40, 0.20]])
prediction = new_observation @ beta
print(f"Predicted target = {prediction[0]:.4f}")
```

### Conclusion
Multiple regression uses several explanatory variables simultaneously. The normal-equation approach provides a direct closed-form solution for the coefficient vector.
